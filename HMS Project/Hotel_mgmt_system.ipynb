#### PROJECT OVERVIEW

**Hotel Management System**

The Hotel Management System is a database-driven application developed using Python and MySQL. It is designed to manage hotel operations such as customer registration, room management, bookings, payments, check-in, check-out and hotel reporting.

The system uses Python for application logic and MySQL for storing and managing data. The two technologies are connected using the mysql.connector library.

#### PROBLEM STATEMENT

Managing hotel operations manually can be time-consuming and may lead to errors in maintaining customer records, room availability, booking details and payment information.

Manual processes make it difficult to track reservations, manage room occupancy, calculate bills and generate business reports efficiently.

The Hotel Management System aims to overcome these challenges by providing a centralized, database-driven solution that automates essential hotel operations using Python and MySQL.

#### PROJECT OBJECTIVES

The main objectives of the Hotel Management System are:

- Customer Management: Maintain customer details and perform customer registration, searching, updating and deletion.

- Room Management: Manage room details, room types, pricing, capacity and availability status.

- Booking Management: Create, view, search and cancel room bookings while preventing overlapping reservations.

- Payment Management: Calculate booking bills, record payments and maintain payment history.

- Check-in and Check-out Management: Update booking and room statuses during customer check-in and check-out.

- Hotel Reports: Generate reports containing customer statistics, room occupancy, booking details, revenue and popular room types.

- Database Management: Store and retrieve hotel information efficiently using MySQL.

- Python–MySQL Integration: Connect Python application logic with a MySQL database using mysql.connector.

#### DATABASE DESIGN

The Hotel Management System uses a relational database named Hotel_mgmt_system, developed using MySQL.

The database consists of four main tables:

1. Customers

    Stores personal and contact information of hotel customers.
    
    customer_id – Primary Key
    
    name – Customer name
    
    contact – Contact number
    
    email – Email address
    
    address – Residential address

2. Rooms

    Stores information about the hotel rooms.
    
    room_no – Primary Key
    
    room_type – Type of room
    
    price_per_day – Daily room price
    
    capacity – Maximum room capacity
    
    status – Room availability status

3. Bookings

    Stores customer reservation details.
    
    booking_id – Primary Key
    
    customer_id – Foreign Key referencing Customers
    
    room_no – Foreign Key referencing Rooms
    
    check_in – Check-in date
    
    check_out – Check-out date
    
    no_of_days – Duration of stay
    
    booking_status – Current booking status

4. Payments

    Stores payment details associated with bookings.
    
    payment_id – Primary Key
    
    booking_id – Foreign Key referencing Bookings
    
    amount – Payment amount
    
    payment_date – Date and time of payment
    
    payment_method – Payment method
    
    payment_status – Payment status

Database Relationships

- One customer can have multiple bookings.

- One room can be associated with multiple bookings over time.

- One booking can have multiple payment records.

Primary keys uniquely identify records, while foreign keys establish relationships between the tables.

#### MODULES DESCRIPTION

The Hotel Management System consists of the following six modules:

**1. Customer Management**

This module manages customer information. It allows users to:
- Add new customers.
- View all customer records.

**2. Room Management**

This module manages hotel room information. It allows users to:

- Delete room records.

**3. Booking Management**

This module handles customer reservations. It allows users to:
- Book rooms for customers.
- View all bookings.
- Search bookings.
- Cancel bookings.
- View customer booking history.

**4. Payment Management**

This module handles booking-related payments. It allows users to:

**5. Check-in / Check-out Management**

This module manages the customer's stay. It allows users to:

- Update booking status to Checked-out.

This module generates reports to monitor hotel operations, including:
- Total customers and rooms.
- Available and occupied rooms.
- Total and cancelled bookings.
- Total revenue.
- Monthly revenue.
- Most popular room type.
- Room occupancy rate.

### System Workflow

The Hotel Management System follows a structured workflow to manage hotel operations.

**Step 1: Customer Registration**

The user enters customer details, which are stored in the Customers table.

**Step 2: Room Management**

The user checks available rooms based on room type, price and capacity.

**Step 3: Room Booking**

The user selects a customer and an available room, enters check-in and check-out dates, and creates a booking.

The system checks for overlapping reservations before confirming the booking.

**Step 4: Bill Calculation and Payment**

The system calculates the booking amount using the room price per day and the number of days.

The payment details are recorded in the Payments table.

**Step 5: Customer Check-in**

The system verifies the booking and updates:
- Booking status to `Checked-in`.
- Room status to `Occupied`.

**Step 6: Customer Check-out**

After payment verification, the system updates:
- Booking status to `Checked-out`.
- Room status to `Available`, provided no other active booking requires the room.

**Step 7: Hotel Reports**

The system generates reports containing customer information, room availability, booking statistics, revenue and occupancy rate.

### Overall Process

Customer Registration → Room Selection → Booking → Bill Calculation → Payment → Check-in → Check-out → Reports

#### BOOKING MANAGEMENT

```python
# BOOK ROOM FUNCTION
def book_room():
    customer_id = int(input("Enter customer ID: "))
    room_no = int(input("Enter room number: "))

    check_in = input("Enter check-in-date (YYYY-MM-DD): ")
    check_out = input("Enter check-out-date (YYYY-MM-DD): ")

    check_in_date = datetime.strptime(check_in, "%Y-%m-%d").date()
    check_out_date = datetime.strptime(check_out, "%Y-%m-%d").date()

    if check_out_date <= check_in_date:
        print("Check-out date must be greater than check-in date!")
        return

    no_of_days = (check_out_date - check_in_date).days

    # CHECK CUSTOMER
    cursor.execute(
        "SELECT customer_id FROM Customers WHERE customer_id = %s",
        (customer_id,)
    )

    if cursor.fetchone() is None:
        print("Customer not found!")
        return

    # CHECK ROOM AVAILABILITY FOR SELECTED DATES
    query = """
    SELECT booking_id FROM Bookings
    WHERE room_no = %s
    AND booking_status IN ('Booked','Checked-in')
    AND check_in < %s
    AND check_out > %s
    """

    cursor.execute(query, (room_no,check_out_date, check_in_date))

    if cursor.fetchone():
        print("Room is already booked for these dates!")
        return

    # INSERT BOOKING
    query = """
    INSERT INTO Bookings 
    (customer_id,room_no,check_in,check_out,no_of_days)
    VALUES (%s,%s,%s,%s,%s)
    """

    data = (customer_id,room_no,check_in_date,check_out_date,no_of_days)
    cursor.execute(query,data)
    connect.commit()

    print("Room booked successfully!")
    print("Total booking days:",no_of_days)
```

#### PAYMENT MANAGEMENT

```python
# CALCULATE BILL FUNCTION
def calculate_bill():
    booking_id = int(input("Enter booking ID: "))

    query = """
    SELECT r.price_per_day, b.no_of_days
    FROM Bookings b
    JOIN Rooms r
    ON b.room_no = r.room_no
    WHERE b.booking_id = %s
    AND b.booking_status = 'Booked'
    """

    cursor.execute(query,(booking_id,))
    result = cursor.fetchone()

    if result is None:
        print("Booking not found or booking is not active!")
        return

    price_per_day = float(result[0])
    no_of_days = result[1]

    total_amount = price_per_day * no_of_days

    print("\n-------------BILL DETAILS--------------------")
    print("Booking ID:", booking_id)
    print("Price Per Day: ₹", price_per_day)
    print("Number of Days:", no_of_days)
    print("Total Amount: ₹", total_amount)
```

```python
# MAKE PAYMENT FUNCTION
def make_payment():
    booking_id = int(input("Enter booking ID: "))

    # CHECK BOOKING AND CALCULATE BILL
    query = """
    SELECT r.price_per_day, b.no_of_days
    FROM Bookings b
    JOIN Rooms r
    ON b.room_no = r.room_no
    WHERE b.booking_id = %s
    AND b.booking_status = 'Booked'
    """

    cursor.execute(query,(booking_id,))
    result = cursor.fetchone()

    if result is None:
        print("Booking not found or booking is not active!")
        return

    price_per_day = float(result[0])
    no_of_days = result[1]

    amount = price_per_day * no_of_days
    print("Total Amount: ₹",amount)

    # PAYMENT DETAILS
    payment_method = input("Enter payment method (Cash/UPI/Card): ")
    payment_status = "Paid"

    # INSERT PAYMENT
    query = """
    INSERT INTO Payments
    (booking_id, amount, payment_method, payment_status)
    VALUES (%s,%s,%s,%s)
    """

    data = (booking_id, amount, payment_method, payment_status)

    cursor.execute(query,data)
    connect.commit()

    print("Payment successful!")
    print("Amount Paid: ₹",amount)
```

#### CHECK-IN MANAGEMENT

```python
# CHECK-IN FUNCTION
def check_in():
    booking_id = int(input("Enter booking ID: "))

    # CHECK WHETHER BOOKING EXISTS
    query = """
    SELECT room_no, booking_status, check_in, check_out
    FROM Bookings 
    WHERE booking_id = %s
    """

    cursor.execute(query,(booking_id,))
    result = cursor.fetchone()

    if result is None:
        print("Booking not found!")
        return

    room_no = result[0]
    booking_status = result[1]
    check_in_date = result[2]
    check_out_date = result[3]

    # CHECK BOOKING STATUS
    if booking_status == "Cancelled":
        print("This booking has been cancelled!")
        return

    if booking_status == "Checked-in":
        print("Customer has already checked in!")
        return

    if booking_status == "Checked-out":
        print("Customer has already checked out!")
        return

    # CHECK BOOKING DATE
    today = datetime.now().date()

    if today < check_in_date:
        print("Check-in date has not arrived yet!")
        return

    if today >= check_out_date:
        print("Booking period has expired!")
        return

    # UPADTE BOOKING STATUS
    query = """
    UPDATE Bookings
    SET booking_status = 'Checked-in'
    WHERE booking_id = %s
    """

    cursor.execute(query,(booking_id,))

    # UPDATE ROOM STATUS
    query = """
    UPDATE Rooms
    SET status = 'Occupied'
    WHERE room_no = %s
    """

    cursor.execute(query,(room_no,))
    
    connect.commit()

    print("Check-in completed successfully!")
    print("Booking ID: ",booking_id)
    print("Room number: ",room_no)
    print("Room Status: Occupied")
```

#### CHECK-OUT MANAGEMENT

```python
# CHECK-OUT FUNCTION
def check_out():
    booking_id = int(input("Enter booking ID: "))

    # CHECK WHETHER BOOKING EXISTS
    query = """
    SELECT room_no, booking_status
    FROM Bookings
    WHERE booking_id = %s
    """

    cursor.execute(query,(booking_id,))
    result = cursor.fetchone()

    if result is None:
        print("Booking not found!")
        return

    room_no = result[0]
    booking_status = result[1]

    if booking_status == "Cancelled":
        print("This booking has been cancelled!")
        return

    if booking_status == "Checked-out":
        print("Customer has already checked out!")
        return

    # CHECK PAYMENT STATUS
    query = """
    SELECT payment_status 
    FROM Payments
    WHERE booking_id = %s
    """

    cursor.execute(query,(booking_id,))
    payment = cursor.fetchone()

    if payment is None or payment[0] != "Paid":
        print("Payment is pending! Please complete payment first.")
        return

    # UPDATE BOOKING STATUS
    query = """
    UPDATE Bookings
    SET booking_status = 'Checked-out'
    WHERE booking_id = %s
    """

    cursor.execute(query,(booking_id,))

    # CHECK FOR OTHER ACTIVE BOOKINGS
    query = """
    SELECT booking_id
    FROM Bookings
    WHERE room_no = %s
    AND booking_status IN ('Booked', 'Checked-in')
    AND check_in <= CURDATE()
    AND check_out > CURDATE()
    """
    
    cursor.execute(query, (room_no,))
    active_booking = cursor.fetchone()

    if active_booking is None:
        query = """
        UPDATE Rooms
        SET status = 'Available'
        WHERE room_no = %s
        """
        cursor.execute(query, (room_no,))
        print("Room is now available!")
    else:
        print("Room has another active booking.")
        
    connect.commit()

    print("Checkout completed successfully!")
    print("Room Number:",room_no)
```

#### BOOKING HISTORY

```python
# CUSTOMER BOOKING HISTORY
def booking_history():
    query = """
    SELECT 
        c.name,
        b.booking_id,
        r.room_no,
        r.room_type,
        b.check_in,
        b.check_out,
        b.no_of_days,
        b.booking_status,
        COALESCE(SUM(p.amount),0) AS total_amount
    FROM Bookings b
    JOIN Customers c
    ON b.customer_id = c.customer_id
    JOIN Rooms r
    ON b.room_no = r.room_no
    LEFT JOIN Payments p
    ON b.booking_id = p.booking_id
    GROUP BY 
        c.name,
        b.booking_id,
        r.room_no,
        r.room_type,
        b.check_in,
        b.check_out,
        b.no_of_days,
        b.booking_status
    ORDER BY b.booking_id
    """

    cursor.execute(query)
    result = cursor.fetchall()

    if result:
        print("\n----------------CUSTOMER BOOKING HISTORY---------------------------")

        for booking in result:
            print("--------------------------------------")
            print("Customer name: ",booking[0])
            print("Booking ID: ",booking[1])
            print("Room number: ",booking[2])
            print("Room type: ",booking[3])
            print("Check-in Date: ",booking[4])
            print("Check-out Date: ",booking[5])
            print("Number of Days: ",booking[6])
            print("Booking Status: ",booking[7])
            print("Total Amount Paid: ",booking[8])

    else:
        print("No booking history found!")
```

#### HOTEL REPORTS AND ANALYTICS

```python
def hotel_reports():

    # TOTAL CUSTOMERS
    query = "SELECT COUNT(*) FROM Customers"
    cursor.execute(query)
    result = cursor.fetchone()

    print("\n--------HOTEL REPORTS-----------")
    print("Total Customers: ",result[0])

    # TOTAL ROOMS
    query = "SELECT COUNT(*) FROM Rooms"
    cursor.execute(query)
    result = cursor.fetchone()

    print("Total Rooms: ",result[0])

    # AVAILABLE ROOMS
    query = "SELECT COUNT(*) FROM Rooms WHERE status = 'Available'"
    cursor.execute(query)
    result = cursor.fetchone()

    print("Available Rooms: ",result[0])

    # OCCUPIED ROOMS
    query = "SELECT COUNT(*) FROM Rooms WHERE status = 'Occupied'"
    cursor.execute(query)
    result = cursor.fetchone()

    print("Occupied Rooms: ",result[0])

    # TOTAL BOOKINGS
    query = "SELECT COUNT(*) FROM Bookings"
    cursor.execute(query)
    result = cursor.fetchone()

    print("Total Bookings: ",result[0])

    # CANCELLED BOOKINGS
    query = "SELECT COUNT(*) FROM Bookings WHERE booking_status = 'Cancelled'"
    cursor.execute(query)
    result = cursor.fetchone()

    print("Cancelled Bookings: ",result[0])

    # TOTAL REVENUE
    query = "SELECT COALESCE(SUM(amount),0) FROM Payments WHERE payment_status = 'Paid'"
    cursor.execute(query)
    result = cursor.fetchone()

    print("Total Revenue: ",result[0])

    # MONTHLY REVENUE
    query = """
    SELECT YEAR(payment_date),
    MONTH(payment_date),
    SUM(amount)
    FROM Payments
    WHERE payment_status = 'Paid'
    GROUP BY YEAR(payment_date), MONTH(payment_date)
    ORDER BY YEAR(payment_date), MONTH(payment_date)
    """
    cursor.execute(query)
    result = cursor.fetchall()

    print("\n--------- MONTHLY REVENUE ---------")

    if result:
        for row in result:
            print("Year: ",row[0], "|Month: ",row[1], "|Revenue: ₹",row[2])
    else:
        print("No revenue records found!")
        
    # MOST POPULAR ROOM TYPE
    query = """
    SELECT r.room_type,
    COUNT(b.booking_id) AS total_bookings
    FROM Rooms r
    JOIN Bookings b
    ON r.room_no = b.room_no
    WHERE b.booking_status != 'Cancelled'
    GROUP BY r.room_type
    ORDER BY total_bookings DESC
    LIMIT 1
    """
    
    cursor.execute(query)
    result = cursor.fetchone()
    print("\n--------------MOST POPULAR ROOM TYPE-------------")

    if result:
        print("Room Type:", result[0])
        print("Total Bookings:", result[1])
    else:
        print("No booking records found!")

    # OCCUPANCY RATE
    query = """
    SELECT COUNT(CASE WHEN status = 'Occupied' THEN 1 END) * 100.0
    / NULLIF(COUNT(*),0)
    FROM Rooms
    """

    cursor.execute(query)
    result = cursor.fetchone()

    print("Occupancy Rate: ",round(result[0] or 0,2),"%")
```
